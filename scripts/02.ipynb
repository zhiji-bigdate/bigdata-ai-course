## 一、注释：写给人看的说明

代码是给电脑执行的，**注释是写给人看的**，电脑会忽略它们。

- **单行注释**：以 `#` 开头
- **多行注释**：用三个引号 `"""..."""` 包起来

```python
# 这是单行注释：电脑不执行这一行，只执行下面的 print
print('Hello, Python!')  # 行尾也可以写注释
```

```python
"""
这是多行注释，
可以一次写很多行，
用来解释一整段代码的作用。
"""
print('多行注释不影响程序运行')
```

## 二、算术运算符：`/`、`//`、`%`

| 运算符 | 名字 | 例子 | 结果 |
| --- | --- | --- | --- |
| `/` | 除法（带小数） | `8 / 3` | 2.666... |
| `//` | 整除（只留整数部分） | `8 // 3` | 2 |
| `%` | 取余（除完剩下多少） | `8 % 3` | 2 |

```python
print(8 / 3)   # 除法：结果带小数
```

```python
print(8 // 3)  # 整数除法，结果为2
```

```python
print(8 % 3)   # 取余数
```

## 三、字符串：用 `[ ]` 取出字符

字符串的每个字符都有编号（索引），**从 0 开始数**。

`"兰州牛肉面"` → 兰=0，州=1，牛=2，肉=3，面=4

```python
s = "兰州牛肉面"
print(s[0])   # 输出第一个字符 '兰'
```

```python
s2 = "牛肉面"    # 新建一个字符串变量
print(s2)
print(s[0] + s2) # 字符串可以用 + 拼接
```

```python
print(s[-1])    # 负索引：倒数第一个字符
print(len(s))   # 字符串长度
```
